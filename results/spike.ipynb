## Spike Dataset

## Imports

```python
from pathlib import Path 
import mne 
import numpy as np 
import pyvista as pv 
import nibabel as nib 
import mindmap as mm

config = mm.io.read_config("../configurations/spike.yml")
```

## Data

```python
sub = "sub-pat10"

# files
mm_dir    = Path(config["MM_DIR"])
resection = mm_dir/sub/"resec"/f"{sub}_auto-seg.nii.gz"
sloreta   = mm_dir/sub/"inv"/f"{sub}_sloreta-0.001"
cset      = mm_dir/sub/"inv"/f"{sub}_cset-0.1"
sources   = mm_dir/sub/"src"/f"{sub}_surf-7mm-src.fif"

# load
resection = nib.load(resection)
sloreta   = mne.read_source_estimate(sloreta, sub)
cset      = mne.read_source_estimate(cset, sub)
sources   = mne.read_source_spaces(sources)

# pre-processing
resection = nib.affines.apply_affine(resection.affine, np.argwhere(resection.get_fdata()>0))/1000
resection = pv.PolyData(resection)

spaces = []
for i in range(len(sources)):
    mask = sources[i]["inuse"].astype(bool)
    spaces.append(sources[i]["rr"][mask])

sources = np.concatenate(spaces)
sources = pv.PolyData(sources)
```

```python
sloreta_loc = np.unravel_index(np.argmax(sloreta.data), sloreta.data.shape)[0]
source_location = pv.PolyData(sources.points[sloreta_loc])

plotter = pv.Plotter()
plotter.add_mesh(sources, point_size=8, color="gray", render_points_as_spheres=True)
plotter.add_mesh(source_location, point_size=15, color="red", render_points_as_spheres=True)
plotter.add_mesh(resection, point_size=1, color="purple")
plotter.show()
```

```python
cset_loc = np.unravel_index(np.argmax(cset.data), cset.data.shape)[0]
source_location = pv.PolyData(sources.points[cset_loc])

plotter = pv.Plotter()
plotter.add_mesh(sources, point_size=8, color="gray", render_points_as_spheres=True)
plotter.add_mesh(source_location, point_size=15, color="red", render_points_as_spheres=True)
plotter.add_mesh(resection, point_size=1, color="purple")
plotter.show()
```
