# Compute the Connectome Spectrum Transform

For every connectome (which can be represented as a graph), there is an underlying connectome spectrum. In mathematical terms, the analysis involves computing the Graph Fourier Transform (GFT).

The GFT uses the eigenvectors of the Graph Laplacian matrix (often the Normalized Graph Laplacian) as its basis functions. These eigenvectors form the spectrum.
- The eigenvectors of the Graph Laplacian correspond to the modes or harmonics on the graph, analogous to the sine and cosine waves of the standard Euclidean Fourier Transform.
- Each eigenvector is associated with a corresponding eigenvalue. This eigenvalue represents the spatial frequency or the measure of variation of the eigenvector across the graph (i.e., how quickly the eigenvector changes value between connected nodes).

## 📦 Imports

```python
import pygsp
import mindmap as mm
```

## 🧠 Data

📥 **<font color=pink>Inputs</font>**
- `sc`: Structural connectome
- `src`: Source space (for visualization)

🎚️ **<font color=pink>Configuration</font>**:
- `fconfig`: Configuration file.
- `sub`: subject id

```python
# ------------ Parameters -------------
fconfig = "../../papers/simf/configuration.yml"
sub     = "10"
# -------------------------------------

reader = mm.io.dataset_reader(fconfig)
data   = reader(sub, "src", "sc")
```

## ⚙️ Connectome Spectrum Transform

📥 **<font color=pink>Inputs</font>**

- `con`: Connectome

📥 **<font color=pink>Outputs</font>**

- `eigvec`: Eigenvectors/harmonics (Transpose of this matrix is the connectome spectrum transform)
- `eigval`: Eigenvalues/frequencies

🎚️ **<font color=pink>Connectome information</font>**:
- `binary`: bool, whether to take the binary or weighted connectome.

```python
# ----------- Parameters -----------
binary = True
# ----------------------------------

con = data["sc"] > 0 if binary else data["sc"]

G = pygsp.graphs.Graph(con, lap_type="normalized")
G.compute_fourier_basis()

eigvec = G.U
eigval = G.e
```

📊 **Harmonics** <br>

```python
# ----------- Parameters -----------
freqs = [0, 1, 2, 3, 4]
# ----------------------------------

mm.viz.plot_harmonics(data["src"], harmonics = eigvec, frequencies = freqs).show()
```

💾 **Save eigenvectors and eigenvalues**

```python
# Save eigendecomposition (eigvec + eigval)
file = mm.io.get_bidspath(
    root        = reader.config["MM_DIR"],
    subject     = data["sub"],
    datatype    = "eig",
    description = "binary" if binary else None,
)
mm.io.mkdir(file)
mm.io.write_eigendecomposition(file, eigvec, eigval)True
print(f"\033[92mSuccessfully saved eigenvectors and eigenvalues {'from binary sc ' if binary else ''}for subject '{data['sub']}'.")
```
