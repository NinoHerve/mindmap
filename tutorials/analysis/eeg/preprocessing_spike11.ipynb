# EEG Preprocessing spike10
This notebook processes EEG data from the SPIKE dataset. The recordings capture intericatly spikes extracted from resting-state EEG patients with epilepsy, with each segment spanning 1 second centered on the spike peak. The data was already averaged.

## 🧪 - Experiment setup
- `eeg`: 'raw' EEG.

```python
%matplotlib widget
import mindmap as mm
import mne
import sys

# reader
dm = mm.io.TutoManager("../../config.yml")
sr = dm.get_subject("spike11")

# data
eeg = sr.load("eeg")
```

## 🧬 - Data Processing

```python
# Average referencing
print("\033[95mSet average referencing ...\033[0m")
eeg = eeg.set_eeg_reference(ref_channels="average", projection=True)
sys.stdout.flush()  # prints the MNE warning here rather then at the end of code

# Baseline correction
print("\033[95mApply baseline correction ...\033[0m")
eeg = eeg.apply_baseline(baseline=(None, 0.25))
```

Make identity matrix as noise covariance matrix.

```python
noise_cov = mne.make_ad_hoc_cov(eeg.info)
```

## 📊 - Visualization

There are many aspects of EEG data to visualize. Let’s generate an MNE report so that all plots can be explored in a web browser.

```python
mm.viz.create_pipeline_report(eeg, noise_cov, subject_id = sr.sub)
```
