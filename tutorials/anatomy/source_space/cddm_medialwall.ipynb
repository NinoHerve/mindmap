# Cortical Distributed Dipole Modeling with Medial Wall Masking

The number of dipoles is adjustable based on the spacing attribute. We then eliminate any dipole that belongs to the medial wall, since these are not associated with gray matter. While the classic MNE-Python pipeline conserves these dipoles, in our case, non-gray matter areas clash with our connectome construction.

## 🧪 - Experiment setup
- `vertex-ctx`: DataFrame of vertices associated to gray matter according to FreeSurfer.

```python
# imports
import mindmap as mm
import mne

# readers
dm = mm.io.TutoManager("../../config.yml")
sr = dm.get_subject("spike11")

# data
vertex_ctx = sr.load("vertex-ctx")
```

## 🧬 - Data Processing

Build Source Spaces

```python
# Setup source space
print("\033[95mSetup source space...\033[0m")
source_space = mne.setup_source_space(
    subject      = f"sub-{sr.sub}",
    spacing      = "oct6",                  # dipole spacing ('octN' is a mesh downsampling method)
    surface      = "white",                 # cortical mesh to use (white or pial)
    add_dist     = True,                    # geodesic distances between sources
    subjects_dir = dm.config["DIR_FS"],
    n_jobs       = 10,                      # number of threads
)

# mask medial wall
print("\033[95mMask medial wall...\033[0m")
source_space = mm.anatomy.source_space.filter_dipoles_by_id_list(source_space, vertex_ctx)

# summary
print("\n\033[92mSUMMARY\033[0m")
for src in source_space:
    print(f"Source space id: {src['id']}")
    print(f"Type: {src['type']}")
    print(f"Number of sources: {src['nuse']}\n")
```

## 📊 - Visualization

```python
mm.viz.plot_brain(source_space, mesh_params={"show_src":True}).show()
```
