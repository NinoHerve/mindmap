# Sensor Coregistration - spike11

To properly compute the forward solution, the sensors (electrode channels or magnetic coils) must be aligned to the head model. <br>

## 🧪 - Experiment setup
- `ave`: Averaged sensors.
- `bem`: BEM layers

```python
import mindmap as mm 
import mne

# readers
dm = mm.io.TutoManager("../../config.yml")
sr = dm.get_subject("spike11")

# data
ave = sr.load("ave", task="rest")
bem = sr.load("bem")
```

## 📊 - Visualization

```python
mm.viz.plot_alignment(bem, ave).show()
```

The EEG channels were already coregistered in the spike dataset!
