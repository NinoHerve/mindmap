# Sensor Coregistration - simf10

In the simulated dataset, we use a template electrode montage that we co-register to the head by aligning it and projecting it onto the head surface.

## 🧪 - Experiment setup
- `eeg`: EEG template model.
- `bem`: Boundary Element Model surfaces.

```python
import mindmap as mm

# readers
dm = mm.io.TutoManager("../../config.yml")
sub = "simf10"

# data
eeg_temp = mm.analysis.functional.eeg.build_eeg_template()       # template EEG
bem = dm.load(sub, "bem")
```

## 🧬 - Data Processing
Coregister electrodes.

```python
eeg_coreg = mm.anatomy.forward.template_coregistration(eeg_temp, f"sub-{sub}", dm.roots["fs"], iterations=0)
```

## 📊 - Visualization

```python
mm.viz.plot_alignment(bem["skin"], eeg_coreg).show()
```
