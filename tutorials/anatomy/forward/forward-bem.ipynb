# Compute Forward Operator - BEM

Compute Forward solution (OpenMEEG via MNE-Python) using a Boundary Element Model (BEM).

## 🧪 - Experiment setup
- `src`: Source spaces
- `ave`: Average recording
- `bem`: BEM layers

```python
# imports
import mindmap as mm 
import mne

# readers
dm = mm.io.TutoManager("../../config.yml")
sr = dm.get_subject("spike11")

# data
ave = sr.load("ave", task="rest")
trans = sr.load("eeg-trans")
src = sr.load("src")
bem = sr.load("bem")
```

## 🧬 - Data Processing

Make forward solution.

```python
forward = mne.make_forward_solution(
    info = ave.info,
    trans = trans,
    src = src,
    bem = bem,
    meg = False,
    eeg = True,
    verbose = "warning",
)
```

## 📊 - Visualization

Check sensitivity

```python
mm.viz.plot_fwd_sensitivity(forward).show()
```

Check source correlation (subsample)

```python
mm.viz.plot_fwd_correlation(forward)
```
