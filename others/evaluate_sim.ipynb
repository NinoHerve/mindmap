# Evaluate Source Reconstruction Applied on Simulated Data

## 📦 Imports

```python
import mindmap as mm
import numpy as np
```

## 🧠 Data

📥 **<font color=pink>Inputs</font>**
- Forward solution (`fwd`): for visualizations (contains source space)
- Connectome spectrum transform (`cst`): To evaluate connectome spectrum 
- Ground truth (`gt`)
- Source Reconstruction (`cset`/`sloreta`/`wcset`/...)

```python
# configuration
config = mm.io.read_config("../configurations/simf.yml")
sub    = config["SUBS"][0]
sample = config["SAMP"][10]

# data
data = mm.io.read_dataset(sub, "fwd", "cst", "eeg_inv", "gt", "cset", "wcset", "enet", "sloreta", sample=sample)

# features
sfreq = data["eeg_inv"].info["sfreq"]

_, ht = mm.utils.eeg_rising_phase(data["eeg_inv"], ratio=0.5)
_, pt = mm.utils.eeg_rising_phase(data["eeg_inv"], ratio=1)

t_half = ht/sfreq 
t_peak = pt/sfreq

print(f"50% rising phase time: {t_half:03}")
print(f"Peak time: {t_peak:03}")
```

## Visualize Results

### 🌟 Visualize Ground Truth$

```python
gt = np.abs(data["gt"].data[:,data["gt"].time_as_index(t_half)])
mm.viz.plot_brain_signal(data["fwd"]["src"], gt, cmap="reds").show()
```

### 🌟 Visualize CSET

```python
cset = np.abs(data["cset"].data[:,data["cset"].time_as_index(t_half)])
mm.viz.plot_brain_signal(data["fwd"]["src"], cset, cmap="reds").show()
```

### 🌟 Visualize sLORETA

```python
sloreta = np.abs(data["sloreta"].data[:,data["sloreta"].time_as_index(t_half)])
mm.viz.plot_brain_signal(data["fwd"]["src"], sloreta, cmap="reds").show()
```

### 🌟 Visualize wCSET

```python
wcset = np.abs(data["wcset"].data[:,data["wcset"].time_as_index(t_half)])
mm.viz.plot_brain_signal(data["fwd"]["src"], wcset, cmap="reds").show()
```

### 🌟 Visualize eNET

```python
enet = np.abs(data["enet"].data[:,data["enet"].time_as_index(t_half)])
mm.viz.plot_brain_signal(data["fwd"]["src"], enet, cmap="reds").show()
```
