# Just play around with code

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng()

n = 1000
size = 100000
fusion = np.zeros(size)
for _ in range(n):
    dist = rng.uniform(5, 10, size=size)
    fusion += dist
```

```python
plt.hist(fusion, bins=100)
plt.show()
```
