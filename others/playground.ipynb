```python
import mindmap as mm 
from pathlib import Path 
import numpy as np
import os
from mindmap.utils.misc import _retrieve_path
import scipy.io
```

```python
mm.io.read_config("../configurations/simf.yml", env=True)
```

```python
def _read_dataset_simf(subject, *args):
    dirs    = _retrieve_path("TVB_DIR", "MM_DIR", "ANA_DIR")
    ana_dir = Path(dirs["ANA_DIR"])
    tvb_dir = Path(dirs["TVB_DIR"])
    mm_dir  = Path(dirs["MM_DIR"])

    data = dict()

    for name in args:
        if name == "ctx":
            # cortical surface
            file = ana_dir/"fs_cortex_20k.mat"
            surf = scipy.io.loadmat(file)
            pos, tri, ori = [surf[key] for key in ["pos", "tri", "ori"]]

            data[name] = mm.utils.surf_to_polydata(pos, tri)
            data[name]["orientation"] = ori

        if name == "fwd": 
            file = ana_dir/"leadfield_75_20k.mat"
            data[name] = scipy.io.loadmat(file)["fwd"]

        if name == "elec":
            file = ana_dir/"electrode_75.mat"
            elec = scipy.io.loadmat(file)
            pos  = [[e[0], e[1], e[2]] for e in elec["eloc75"][0,:]]
            pos  = np.array(pos).squeeze()

            data[name] = pos

    return data
```

```python
data = _read_dataset_simf(None, "ctx", "fwd", "elec")
```

```python
import pyvista as pv 

plotter = pv.Plotter() 
plotter.add_mesh(data["ctx"], color="gray", opacity=0.1)
plotter.add_mesh(pv.PolyData(data["elec"]), color="lightblue")
plotter.show()
```
